### Imports 

```python
from helpers import load_csv_dataset
from pathlib import Path

data_dir = Path('../data')
respondents_dataset = load_csv_dataset(data_dir.joinpath('respondents.csv'), delimiter = ';')
respondents_dataset = [x for x in respondents_dataset if x['valid'] == '1']
scenarios = ['s1', 's2', 's3', 's4', 's5']
```

### Demographics

#### 1. Number of respondents per scenario

```python
scenarios = sorted(set([x['scenario'] for x in respondents_dataset]))

for x in scenarios:
    count = 0
    for respondent in respondents_dataset:
        count = count + 1 if respondent['scenario'] == x else count + 0
    print(f'{x}: {count}')
```

#### 2. Number of respondents per role

```python
roles = set([x['role'] for x in respondents_dataset])
table_latex = ''
for x in roles:
    table_latex += x
    count_s = []
    for s in scenarios:
        count = 0
        for respondent in respondents_dataset:
            count = count + 1 if (respondent["role"] == x) and (respondent["scenario"] == s) else count + 0
        count_s.append(count)    
    print(f'{x}: {count_s} => {sum(count_s)}')
```

#### 3. Number of respondents per experience

```python
roles = set([x['experience'] for x in respondents_dataset])
for x in roles:
    count_s = []
    for s in scenarios:
        count = 0
        for respondent in respondents_dataset:
            count = count + 1 if (respondent["experience"] == x) and (respondent["scenario"] == s) else count + 0
        count_s.append(count)
        
    print(f'{x}: {count_s} => {sum(count_s)}')
```

#### 4. Number of respondents per education level

```python
roles = set([x['education'] for x in respondents_dataset])

for x in roles:
    count_s = []
    for s in scenarios:
        count = 0
        for respondent in respondents_dataset:
            count = count + 1 if (respondent["education"] == x) and (respondent["scenario"] == s) else count + 0
        count_s.append(count)
        
    print(f'{x}: {count_s} => {sum(count_s)}')
```

#### 5. Perceived usefulness of the scenarios

```python
respondents_dataset = load_csv_dataset(data_dir.joinpath('usefulness.csv'), delimiter = ';')
respondents_dataset = [x for x in respondents_dataset if x['valid'] == '1']
scenarios = ['s1', 's2', 's3', 's4', 's5']
usefulness = set([x['usefulness'] for x in respondents_dataset])

for x in usefulness:
    count_s = []
    for s in scenarios:
        count = 0
        for respondent in respondents_dataset:
            count = count + 1 if (respondent["usefulness"] == x) and (respondent["\ufeffscenario"] == s) else count + 0
        count_s.append(count)
        
    print(f'{x}: {count_s} => {sum(count_s)}')
```
